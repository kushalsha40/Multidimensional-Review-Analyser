```python
from nltk.tokenize import word_tokenize, sent_tokenize
import re
from nltk.corpus import stopwords , wordnet 
import nltk
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
import string
```

```python
lemmatizer = WordNetLemmatizer()


def pipeline(text):
    # Remove HTML tags, URLs, mentions, hashtags, and non-alphabetic characters
    text = re.sub(r'<.*?>' ,' ', text)
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    text = re.sub(r'http\S+|www\.\S+', ' ', text)
    text = re.sub(r'[@#]\w+', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()

    # Preprocesing stpes handling negation words and stop words
    text = text.lower()
    word_tokens = word_tokenize(text)

    # defining the trigger words for negation handling
    negation_words = {'not' , 'no' , 'never' , 'nothing' , 'nowhere' , 'neither' , 'nor' , 'cannot' , "can't" , "don't" , "doesn't" , "didn't" , "won't" , "wouldn't" , "shouldn't" , "isn't" , "aren't" , "wasn't" , "weren't"}
    stop_words = set(stopwords.words('english'))

    transformed = []
    skip_next = False

    for i in range(len(word_tokens)):
        word = word_tokens[i]
        if skip_next:
            skip_next = False
            continue
     
        if word in negation_words and (i+1) < len(word_tokens):
            next_word = word_tokens[i+1]
            # Linguistic Accuracy: Stemming might turn "was" into "wa," but Lemmatization turns "was" into "be."
            lemma_next = lemmatizer.lemmatize(next_word , pos=wordnet.VERB)  # I used pos = wordnet.VERB. This tells the lemmatizer to treat the word as a verb.
            combined = f"{word}_{lemma_next}" # I lematize the word before combining it with the negation word. This way, we capture the base form of the verb while still indicating that it's negated.
            transformed.append(combined)
            skip_next = True
        
        elif word not in stop_words and word not in string.punctuation:
            lemma_word = lemmatizer.lemmatize(word , pos = wordnet.VERB)
            transformed.append(lemma_word)
        
    return ' '.join(transformed)


data['review'] = data['review'].apply(pipeline)
print(data['review'].head())
```
